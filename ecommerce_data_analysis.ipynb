merging the datsets

```python
# Aggregate items per order
items_agg = items.groupby('order_id').agg(
    total_items=('order_item_id', 'count'),
    revenue=('price', 'sum'),
    freight=('freight_value', 'sum')
).reset_index()

# Aggregate payments per order
payments_agg = payments.groupby('order_id').agg(
    payment_value=('payment_value', 'sum'),
    installments=('payment_installments', 'max'),
    payment_type=('payment_type', lambda x: x.mode()[0])  # most used type
).reset_index()

# Now merge safely
df = pd.merge(orders, items_agg, on='order_id', how='inner')
df = pd.merge(df, payments_agg, on='order_id', how='inner')
```

```python
df.isnull().sum()
df.duplicated().sum()
```

filtering the data to "delivered" because we want to consider only complete transactions

```python
# 1. Convert timestamps
for col in ['order_purchase_timestamp', 'order_delivered_customer_date',
            'order_estimated_delivery_date']:
    df[col] = pd.to_datetime(df[col])

# 2. Filter only completed orders — cancelled/unavailable orders skew revenue
df = df[df['order_status'] == 'delivered']

# 3. Drop rows with null revenue (can't analyze what we can't measure)
df.dropna(subset=['revenue'], inplace=True)
```

data transformation

```python
# Time-based features — essential for trend analysis
df['order_month'] = df['order_purchase_timestamp'].dt.to_period('M')
df['order_year'] = df['order_purchase_timestamp'].dt.year
df['order_quarter'] = df['order_purchase_timestamp'].dt.to_period('Q')
df['day_of_week'] = df['order_purchase_timestamp'].dt.day_name()

# Revenue per item (avg order value insight)
df['avg_item_price'] = df['revenue'] / df['total_items']

# Freight ratio — what % of revenue is freight?
df['freight_ratio'] = df['freight'] / (df['revenue'] + df['freight'])
```

univariate analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Revenue distribution — expect heavy right skew
sns.histplot(df['revenue'], bins=50, log_scale=True)
plt.title('Revenue Distribution (log scale)')
plt.show()

# Payment type breakdown
df['payment_type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Payment Type Distribution')
plt.show()

# Orders per month
df.groupby('order_month')['order_id'].count().plot(figsize=(14,4))
plt.title('Monthly Order Volume')
plt.show()
```

bivariate analysis

```python
# Monthly revenue trend (the most important plot)
monthly_rev = df.groupby('order_month')['revenue'].sum()
monthly_rev.plot(figsize=(14,4), marker='o')
plt.title('Monthly Revenue Over Time')
plt.ylabel('Total Revenue (BRL)')
plt.show()

# Revenue by payment type
df.groupby('payment_type')['revenue'].mean().sort_values().plot(kind='barh')
plt.title('Avg Order Revenue by Payment Type')
plt.show()

# Installments vs revenue — do people pay in installments for bigger purchases?
sns.boxplot(x='installments', y='revenue', data=df[df['installments'] <= 12])
plt.title('Revenue vs Number of Installments')
plt.show()

# Day of week ordering pattern
df['day_of_week'].value_counts().reindex(
    ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
).plot(kind='bar')
plt.title('Orders by Day of Week')
plt.show()
```

correlation anaysis

```python
# Select meaningful numeric columns
corr_cols = df[['revenue', 'freight', 'total_items', 'installments',
                'payment_value', 'avg_item_price', 'freight_ratio']]

sns.heatmap(corr_cols.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap — Revenue & Order Features')
plt.show()
```

visualization

```python
# Quarter-over-quarter revenue growth
qtr_rev = df.groupby('order_quarter')['revenue'].sum()
qtr_growth = qtr_rev.pct_change() * 100
qtr_growth.plot(kind='bar', color='steelblue')
plt.title('Quarter-over-Quarter Revenue Growth (%)')
plt.axhline(0, color='red', linestyle='--')
plt.show()

# Revenue vs freight over time (dual axis)
fig, ax1 = plt.subplots(figsize=(14,5))
monthly_rev.plot(ax=ax1, color='steelblue', label='Revenue')
ax2 = ax1.twinx()
df.groupby('order_month')['freight'].sum().plot(ax=ax2, color='orange', label='Freight')
plt.title('Revenue vs Freight Cost Over Time')
fig.legend(loc='upper left')
plt.show()
```
